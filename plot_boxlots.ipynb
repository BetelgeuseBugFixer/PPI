```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
scores_oh_easy = pd.read_pickle("dataset/scores/scores_oh_easy_all.pkl")
scores_oh_hard = pd.read_pickle("dataset/scores/scores_oh_hard_all.pkl")
```

```python
scores_emb_easy = pd.read_pickle("dataset/scores/scores_emb_easy_all.pkl")
scores_emb_hard = pd.read_pickle("dataset/scores/scores_emb_hard_all.pkl")
```

```python
metrics=["F1-score", "Precision", "Recall", "Balanced Accuracy", "FPR", "Accuracy"]
color_1 = 'purple'
color_2 = 'lightblue'
fontsize=18

plt.figure(figsize=(12, 6))

positions = []
labels = []
data = []
colors = []

# Generate positions and labels for boxplots
for i, metric in enumerate(metrics):
    positions.extend([i * 1.5 + 1, i * 1.5 + 1.6])  # Slightly more spacing between easy and hard
    labels.append(metric)
    data.append(scores_oh_hard[metric])
    data.append(scores_emb_hard[metric])
    colors.extend([color_1, color_2])

box = plt.boxplot(data, patch_artist=True, positions=positions)

for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

plt.xticks(ticks=[i * 1.5 + 1.3 for i in range(len(metrics))], labels=labels, rotation=0, fontsize=fontsize)
plt.title("Hard dataset",fontsize=fontsize)
plt.ylabel("Score",fontsize=fontsize)
plt.grid(True)
plt.legend([box["boxes"][0], box["boxes"][1]], ["Onehot-Encoded Model", "Embedding Model"], loc='right', fontsize=fontsize)
plt.tight_layout()
plt.show()
```
