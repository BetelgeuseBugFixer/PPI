```python
import numpy as np
import pandas as pd
```

```python
# load data
df = pd.read_pickle("./dataset.pkl")
```

```python
df["length"] = df["sequence"].str.len()
```

```python
# run this if working with subset of dataset
df = df[df['length'] <= 500].head(10000)
```

```python
# pad sequences and lists

def pad_list(tensor_list, max_len):
    return tensor_list + [None] * (max_len - len(tensor_list))

def pad_sequence(sequence, max_len):
    return sequence + "-"*(max_len - len(sequence))


df["padded_pfam_tensor"] = df["pfam_tensor"].apply(
    lambda x: pad_list(x, 500)
)

df["padded_sequence"] = df["sequence"].apply(
    lambda x: pad_sequence(x, 500)
)

df
```

```python
# one hot encode both sequences and lists


unique_pfams = set()
for pfam in df['padded_pfam_tensor'].values:
    unique_pfams.update(pfam)

unique_pfams = list(unique_pfams)

unique_pfams.remove(None)
unique_pfams.insert(0, None)

unique_pfams
```

```python
PROTEIN_ALPHABET = {
    '-': 0,
    'A': 1,
    'C': 2,
    'D': 3,
    'E': 4,
    'F': 5,
    'G': 6,
    'H': 7,
    'I': 8,
    'K': 9,
    'L': 10,
    'M': 11,
    'N': 12,
    'P': 13,
    'Q': 14,
    'R': 15,
    'S': 16,
    'T': 17,
    'V': 18,
    'W': 19,
    'Y': 20,
    'X': 21
}
```

```python
def one_hot_pfam(pfam_list):
    return [unique_pfams.index(pfam) for pfam in pfam_list]

def one_hot_sequence(sequence):
    return [PROTEIN_ALPHABET[aa] for aa in sequence]


df["pfam_oh_index"] = df["padded_pfam_tensor"].apply(
    lambda x: one_hot_pfam(x)
)

df["sequence_oh_index"] = df["padded_sequence"].apply(
    lambda x: one_hot_sequence(x)
)

df
```

```python
def one_hot(indices, one_hot_len):
    vec = np.zeros(one_hot_len, dtype=int)
    vec[indices] = 1
    return vec
```
